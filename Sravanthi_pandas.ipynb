Q1. Two NumPy arrays have shapes (3, 1) and (1, 3). Predict the shape of their element-wise sum and explain how broadcasting works. Why might the result differ from adding two one-dimensional arrays?

import numpy as np

a = np.array([[10], [20], [30]])

b = np.array([[1, 2, 3]])

result = a + b

print(result)

print(result.shape)

1: Predict the output of the given code.


```python
import numpy as np

a = np.array([[10], [20], [30]])
b = np.array([[1, 2, 3]])

result = a + b

print(result)
print(result.shape)
```

2: Identify the shapes of arrays a and b before addition.

Ans:

Shape of a: (3, 1)

Shape of b: (1, 3)

3:  Explain how NumPy broadcasting enables the addition of these arrays.

Ans:

 Broadcasting automatically expands the arrays to compatible shapes without explicitly copying their values. Array a is expanded across columns, and array b is expanded across rows, producing a (3, 3) result.

 What would happen if b were replaced with np.array([1, 2])? Explain the reason.

Ans:

 A ValueError occurs because the shapes (3, 1) and (2,) are not compatible for broadcasting. Their dimensions do not match and neither dimension is 1.

 Explanation: 
 
 Reshaping b to (1, 3) makes the arrays compatible for broadcasting.

```python
b = np.array([1, 2, 3]).reshape(1, 3)
result = a + b

print(result)
print(result.shape)
```

Given dataset: students.csv

Given code:

1: Predict how the quoted comma in "Asha, R" will be handled when the CSV file is read.

Ans:

 Pandas correctly treats "Asha, R" as a single field because the comma is inside double quotes. It will not split the name into two columns.

2: Identify the potential issues with the Date and Marks columns.

Ans:

Date: Dates are in different formats, such as 2026-09-01 and 01/09/2026, which may cause inconsistent parsing.

Marks: Blank values and "NA" may represent missing data. They should be checked and converted appropriately.

3: Rewrite the code to handle missing marks and convert the date column into a suitable date-time format.


```python
import pandas as pd
from io import StringIO

data = """ID,Name,Date,Marks
101,"Asha, R",2026-09-01,80
102,Ravi,01/09/2026,
103,Neha,2026-09-03,NA"""

df = pd.read_csv(StringIO(data), na_values=["", "NA"])

df.columns = df.columns.str.strip()

df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")

df["Date"] = pd.to_datetime(
    df["Date"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

print(df)
print(df.dtypes)
```

Q1:Student Assessment Dataset Cleaning Using Pandas

Question: Using Pandas, perform the following cleaning operations in sequence:

-> Load the CSV file and inspect the dataset using head(), info(), and isnull().sum().

-> Standardize department names such as IT, it, and I.T. Apply the same approach to Finance.

-> Identify and remove exact duplicate records.

-> Convert Marks into numeric format, handling "88 marks", "Absent", "NA", and blank entries.

-> Investigate missing marks and distinguish absent students from students whose marks have not been entered. State one risk of deleting missing values without investigating why they are missing.

-> Identify marks outside the valid range of 0–100 and decide how they should be handled.

-> Display the cleaned DataFrame and report the number of records removed, missing marks remaining, and invalid marks identified.

```python
import pandas as pd
from io import StringIO

data = """Student_ID,Student_Name,Department,Assessment,Marks
S101,Asha,IT,Python,85
S102,Ravi,it,Python,78
S103,Neha,I.T.,Python,92
S104,Karan,HR,Python,
S105,Meera,HR,Python,Absent
S106,Arjun,Finance,Python,0
S107,Priya,finance,Python,67
S108,Rahul,IT,Python,75
S102,Ravi,it,Python,78
S109,Sneha,I.T.,Python,NA
S110,Varun,HR,Python,88 marks
S111,Diya,Finance,Python,91
S112,Aman,IT,Python,
S113,Nisha,HR,Python,64
S114,Kabir,Finance,Python,105
"""

df = pd.read_csv(StringIO(data))

df["Department"] = df["Department"].str.strip().str.upper()
df["Department"] = df["Department"].replace("I.T.", "IT")

df = df.drop_duplicates()

df["Marks"] = (
    df["Marks"].astype(str)
    .str.replace("marks", "", case=False)
    .str.strip()
    .replace(["", "NA", "ABSENT", "NAN"], pd.NA)
)

df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")

df.loc[df["Marks"] > 100, "Marks"] = pd.NA

df["Marks"] = df.groupby("Department")["Marks"].transform(
    lambda x: x.fillna(x.median())
)

print("Cleaned Dataset:")
print(df)

print("\nData Types:")
print(df.dtypes)
```

Risk of deleting missing values:

Deleting missing values without investigation may cause loss of important student information and lead to biased results.

Justification:

Pandas is used because it provides simple methods for data cleaning, duplicate removal, missing value handling, and numeric conversion. CSV is suitable for storing student assessment data in tabular form.

Explanation:

-> head(), info(), and isnull().sum() inspect the dataset.

-> replace() standardizes department names.

-> drop_duplicates() removes repeated records.

-> to_numeric() converts marks into numeric values.

-> Absent, NA, and blank entries are treated as missing, but absence is kept distinguishable as a reason for missingness during investigation.

->Marks outside 0–100 are identified and treated as invalid.
